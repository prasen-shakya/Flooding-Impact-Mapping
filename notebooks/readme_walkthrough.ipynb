## 2. Look at it

```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt

n = 12
fig, axes = plt.subplots(n, 3, figsize=(15, 5 * n))   # 5 inches per row

for i in range(1, n + 1):
    scene = f"data/image_{i}"
    row = axes[i - 1]

    with rasterio.open(f"{scene}/image.tif") as src:
        image = src.read()
    with rasterio.open(f"{scene}/dem.tif") as src:
        dem = src.read(1)
    with rasterio.open(f"{scene}/label.tif") as src:
        label = src.read(1)

    rgb = np.transpose(image[:3], (1, 2, 0)) / 255.0
    row[0].imshow(rgb)
    row[0].set_title(f"image_{i}: RGB")

    row[1].imshow(dem, cmap="terrain")
    row[1].set_title(f"image_{i}: DEM")

    # label: 0 = dry (white), 1 = flood (blue), -1 = unknown (red)
    row[2].imshow(np.ma.masked_equal(label, -1), cmap="Blues", vmin=0, vmax=1)
    row[2].set_facecolor("red")
    row[2].set_title(f"image_{i}: label")

    for ax in row:
        ax.set_xticks([])
        ax.set_yticks([])

plt.tight_layout()
plt.savefig("all_scenes_preview.png", dpi=100)
print("Saved all_scenes_preview.png")
```

3. Check the label's class balance

```python
from pathlib import Path
import numpy as np
import rasterio

totals = {0: 0, 1: 0, -1: 0}
for label_path in sorted(Path("data").glob("image_*/label.tif")):
    with rasterio.open(label_path) as src:
        label = src.read(1)
        values, counts = np.unique(label, return_counts=True)
    c = dict(zip(values.tolist(), counts.tolist()))
    f, d = c.get(1, 0), c.get(0, 0)
    print(f"{label_path.parent.name}: {100 * f / (f + d):.1f}% flooded")
    for v, c in zip(values, counts):
        totals[int(v)] = totals.get(int(v), 0) + int(c)

dry, flood, nodata = totals[0], totals[1], totals[-1]
known = dry + flood
print(f"dry: {dry:,}, flood: {flood:,}, nodata: {nodata:,}")
print(f"% flooded (of known pixels): {100 * flood / known:.2f}%")
```
